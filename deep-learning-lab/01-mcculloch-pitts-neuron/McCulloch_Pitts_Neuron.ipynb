# McCulloch-Pitts Neuron

This notebook demonstrates a simple threshold neuron and explains how it implements decision logic.

## Understanding the Problem

The McCulloch-Pitts model is a foundational artificial neuron. It takes binary inputs, multiplies them by weights, sums the values, and compares the result against a threshold.

```python
import pandas as pd

from mcp_neuron import AND_gate, NOR_gate, NOT_gate, OR_gate, XOR_gate, gate_truth_table, single_input_truth_table

binary_inputs = [0, 1]

and_table = gate_truth_table(AND_gate, binary_inputs)
or_table = gate_truth_table(OR_gate, binary_inputs)
nor_table = gate_truth_table(NOR_gate, binary_inputs)
not_table = single_input_truth_table(NOT_gate, binary_inputs)
xor_table = gate_truth_table(XOR_gate, binary_inputs)

print('AND gate truth table:')
print(and_table)
print('\nOR gate truth table:')
print(or_table)
print('\nNOR gate truth table:')
print(nor_table)
print('\nNOT gate truth table:')
print(not_table)
print('\nXOR gate truth table:')
print(xor_table)
```

## Core Concept

The neuron uses the rule:

$Z = \sum_i x_i w_i$

and then applies a threshold. This simple rule explains how logic gates can be represented with a threshold-based artificial neuron.

```python
df_and = pd.DataFrame(and_table, columns=['x1', 'x2', 'output'])
df_or = pd.DataFrame(or_table, columns=['x1', 'x2', 'output'])
df_nor = pd.DataFrame(nor_table, columns=['x1', 'x2', 'output'])
df_not = pd.DataFrame(not_table, columns=['x', 'output'])
df_xor = pd.DataFrame(xor_table, columns=['x1', 'x2', 'output'])

display(df_and)
display(df_or)
display(df_nor)
display(df_not)
display(df_xor)
```

## Why XOR is Different

A single threshold neuron can represent linearly separable functions, but XOR is not linearly separable. In other words, no single straight decision boundary can separate the 1s from the 0s in the XOR truth table, which is why the model requires multiple neurons or a hidden layer.

```python
samples = [(0, 0), (0, 1), (1, 0), (1, 1)]

print('XOR outputs:')
for x1, x2 in samples:
    print(f'({x1}, {x2}) -> {XOR_gate(x1, x2)}')

print('\nA single MCP neuron cannot represent XOR because the 1s are not separable by one linear threshold rule.')
```
